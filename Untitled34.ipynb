```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout
from tensorflow.keras.utils import to_categorical
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)
```

```python
plt.figure(figsize=(10,5))

for i in range(10):

    plt.subplot(2,5,i+1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
from tensorflow.keras.layers import Input

model = Sequential([

    Input(shape=(28,28)),

    Flatten(),

    Dense(128, activation='relu'),

    Dense(64, activation='relu'),

    Dropout(0.3),

    Dense(10, activation='softmax')

])
```

```python
model.compile(

    optimizer='adam',

    loss='categorical_crossentropy',

    metrics=['accuracy']

)
```

```python
model.summary()
```

```python
history = model.fit(

    x_train,
    y_train,

    epochs=7,

    batch_size=32,

    validation_split=0.2

)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')

plt.xlabel('Epoch')

plt.ylabel('Accuracy')

plt.legend(['Training Accuracy', 'Validation Accuracy'])

plt.show()
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('Model Loss')

plt.xlabel('Epoch')

plt.ylabel('Loss')

plt.legend(['Training Loss', 'Validation Loss'])

plt.show()
```

```python
predictions = model.predict(x_test)

plt.figure(figsize=(10,5))

for i in range(10):

    plt.subplot(2,5,i+1)

    plt.imshow(x_test[i], cmap='gray')

    predicted_label = np.argmax(predictions[i])

    plt.title(f"Pred: {predicted_label}")

    plt.axis('off')

plt.tight_layout()

plt.show()
```

```python
model_sgd = Sequential([

    Flatten(input_shape=(28,28)),

    Dense(128, activation='relu'),

    Dense(64, activation='relu'),

    Dropout(0.3),

    Dense(10, activation='softmax')

])
```

```python
model_sgd.compile(

    optimizer='sgd',

    loss='categorical_crossentropy',

    metrics=['accuracy']

)
```

```python
history_sgd = model_sgd.fit(

    x_train,
    y_train,

    epochs=7,

    batch_size=32,

    validation_split=0.2

)
```

```python
test_loss_sgd, test_accuracy_sgd = model_sgd.evaluate(x_test, y_test)

print("\nSGD Test Loss:", test_loss_sgd)
print("SGD Test Accuracy:", test_accuracy_sgd)
```

```python
plt.plot(history.history['val_accuracy'])
plt.plot(history_sgd.history['val_accuracy'])

plt.title('Adam vs SGD Accuracy')

plt.xlabel('Epoch')

plt.ylabel('Validation Accuracy')

plt.legend(['Adam', 'SGD'])

plt.show()
```

```python
plt.plot(history.history['val_loss'])
plt.plot(history_sgd.history['val_loss'])

plt.title('Adam vs SGD Loss')

plt.xlabel('Epoch')

plt.ylabel('Validation Loss')

plt.legend(['Adam', 'SGD'])

plt.show()
```
